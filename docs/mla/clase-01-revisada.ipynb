# 1. Repaso de Machine Learning: álgebra y geometría

Los datos se representan como vectores; los modelos aprenden relaciones a partir de ellos. Esta clase conecta transformaciones afines, activaciones, aprendizaje supervisado y no supervisado y la geometría de las máquinas de vectores de soporte.



*Fig. 1.1. Supervisado y no supervisado utilizan objetivos diferentes.*

## 1.1. El espacio de características

Una observación con $d$ atributos se representa mediante $\mathbf{x}\in\mathbb{R}^d$. Para $n$ observaciones, las filas de $\mathbf{X}\in\mathbb{R}^{n\times d}$ contienen las muestras.

En aprendizaje supervisado buscamos una función $f(\mathbf{x})$ que prediga una etiqueta o un valor. En aprendizaje no supervisado buscamos estructura sin una variable objetivo proporcionada. Las características pueden estar correlacionadas; tener $d$ columnas no significa que existan $d$ factores independientes.



*Fig. 1.2. X tiene n filas y d columnas; cada fila representa una muestra.*

## 1.2. Qué puede representar un modelo lineal

Una capa afín con muestras por filas tiene la forma

$$\mathbf{O}=\mathbf{X}\mathbf{W}+\mathbf{1}\mathbf{b}^{T},\qquad \mathbf{W}\in\mathbb{R}^{d\times h}. $$

Cada puntuación tiene derivada constante respecto de una entrada: $\partial o_j/\partial x_k=W_{kj}$. Esta propiedad se refiere a las **puntuaciones afines**. Después de una función como softmax, las probabilidades ya no son funciones afines y no deben confundirse con esas puntuaciones.

Una relación en forma de U, por ejemplo $y=(x-c)^2$, necesita características transformadas o un modelo no lineal. En imágenes, las interacciones entre píxeles también pueden requerir no linealidad. Estas observaciones no implican que una transformación particular de la imagen conserve siempre su clase.



*Fig. 1.3. Curvas ilustrativas: una función afín no representa por sí sola una relación en U.*

## 1.3. Composición de capas afines

Sin activaciones intermedias,

$$\mathbf{H}=\mathbf{X}\mathbf{W}_1+\mathbf{1}\mathbf{b}_1^T,\qquad \mathbf{O}=\mathbf{H}\mathbf{W}_2+\mathbf{1}\mathbf{b}_2^T.$$

Al sustituir,

$$\mathbf{O}=\mathbf{X}(\mathbf{W}_1\mathbf{W}_2)+\mathbf{1}(\mathbf{b}_1^T\mathbf{W}_2+\mathbf{b}_2^T).$$

La composición sigue siendo afín. Una capa estrecha puede imponer una restricción de rango a la matriz equivalente: la familia representable puede ser incluso menor. Para clasificación binaria, el nivel de puntuación cero es un hiperplano; en multiclase aparecen fronteras por pares. Una transformación afín singular puede colapsar dimensiones, por lo que tampoco conserva necesariamente la topología del espacio.

## 1.4. Activaciones no lineales

Una red introduce, por ejemplo, $\mathbf{H}=\sigma(\mathbf{X}\mathbf{W}+\mathbf{b})$, aplicando $\sigma$ elemento a elemento.

| Activación | Expresión | Propiedad |
|---|---|---|
| ReLU | $\max(0,x)$ | Continua y lineal a trozos; gradiente cero para $x<0$. |
| Sigmoide | $1/(1+e^{-x})$ | Rango $(0,1)$; útil como probabilidad Bernoulli cuando el modelo y la pérdida se formulan para ello. |
| Tanh | $(e^x-e^{-x})/(e^x+e^{-x})$ | Rango $(-1,1)$ y simetría impar; puede saturarse. |

Las activaciones amplían la capacidad de representación. No garantizan que un entrenamiento finito encuentre una representación separable ni que esa representación generalice.



*Fig. 1.4. Las activaciones introducen no linealidad; la separabilidad no está garantizada.*

## 1.5. Aprendizaje supervisado y no supervisado

En supervisado contamos con pares $(\mathbf{x}_i,y_i)$. La **clasificación** predice categorías y la **regresión** predice cantidades. Las etiquetas no garantizan separabilidad perfecta: puede haber ruido, clases solapadas o información insuficiente.



*Fig. 1.5. Clasificación y regresión son tareas supervisadas diferentes.*

### 1.5.1. Modelos paramétricos y no paramétricos

Una red de arquitectura fija tiene un número de parámetros determinado por esa arquitectura. Se ajusta minimizando una pérdida; su representación interna puede ser no lineal.

En $k$ vecinos más cercanos se conservan las muestras de entrenamiento y se predice a partir de sus vecinos según una métrica. La distancia euclídea es

$$d(\mathbf{x},\mathbf{x}')=\sqrt{\sum_{j=1}^{d}(x_j-x'_j)^2}.$$

La escala de los atributos influye en la vecindad. «No paramétrico» no significa que el método carezca de hiperparámetros: $k$, la métrica y la ponderación deben seleccionarse con validación. Para 1-NN, las regiones se relacionan con celdas de Voronoi y pueden describirse matemáticamente.



*Fig. 1.6. El escalado y la métrica modifican quiénes son los vecinos.*

### 1.5.2. Estructura sin etiquetas

PCA busca direcciones de gran varianza; $k$-means agrupa por proximidad a centroides. Son objetivos distintos. Una componente de alta varianza o un grupo geométrico no tiene por qué coincidir con una clase semántica.



*Fig. 1.7. Retener varianza no equivale a identificar clases semánticas.*

#### 1.5.2.1. PCA: centrado y proyección

Para datos centrados $\mathbf{X}_c$, una estimación habitual de la covarianza es

$$\mathbf{S}=\frac{1}{n-1}\mathbf{X}_c^T\mathbf{X}_c.$$

Usar $1/n$ cambia los autovalores, pero no las direcciones principales. PCA selecciona autovectores ortonormales asociados a los mayores autovalores, o equivalentemente utiliza una SVD de los datos centrados.

La proyección a $k$ componentes es $\mathbf{Z}=\mathbf{X}_c\mathbf{V}_k$. Maximiza la varianza retenida y minimiza el error de reconstrucción cuadrático entre proyecciones lineales ortogonales de rango $k$. No asegura conservar las variables útiles para predecir una etiqueta. Estandarizar o no estandarizar cambia el problema.

#### 1.5.2.2. $k$-means y celdas de Voronoi

$k$-means alterna asignaciones al centroide más próximo y actualización de centroides como medias, buscando reducir

$$J=\sum_i\|\mathbf{x}_i-\boldsymbol{\mu}_{z_i}\|^2.$$

Para centroides fijos,

$$V_j=\{\mathbf{x}:\|\mathbf{x}-\boldsymbol{\mu}_j\|\leq\|\mathbf{x}-\boldsymbol{\mu}_l\|\ \forall l\}.$$

Las fronteras entre dos centroides son partes de sus hiperplanos bisectores. Las celdas son convexas, pero pueden ser **no acotadas**. El algoritmo converge a una solución local; conviene probar varias inicializaciones. La métrica euclídea y la escala de los atributos condicionan los grupos.

### 1.5.3. SVM lineal: margen máximo

Para etiquetas $y_i\in\{-1,+1\}$, la decisión se basa en el signo de $\mathbf{w}^T\mathbf{x}+b$. En el caso separable de margen duro, el SVM busca el hiperplano que maximiza la distancia mínima a las muestras de ambas clases. Los puntos que activan las restricciones determinan el margen.



*Fig. 1.8. Las clases de este ejemplo son separables; el margen suave admite violaciones.*

#### 1.5.3.1. Normalización canónica

En el caso separable se puede fijar la escala de $(\mathbf{w},b)$ para exigir

$$y_i(\mathbf{w}^T\mathbf{x}_i+b)\geq1.$$

Los hiperplanos de apoyo tienen puntuaciones $+1$ y $-1$. La distancia desde la frontera central hasta cada uno es $1/\|\mathbf{w}\|$.

#### 1.5.3.2. Optimización del margen y regularización

El ancho total entre los hiperplanos de apoyo es $2/\|\mathbf{w}\|$. El problema de margen duro es

$$\min_{\mathbf{w},b}\frac12\|\mathbf{w}\|^2\quad\text{sujeto a}\quad y_i(\mathbf{w}^T\mathbf{x}_i+b)\geq1.$$

Si los datos no son separables, usamos **margen suave**:

$$\min_{\mathbf{w},b,\boldsymbol{\xi}}\frac12\|\mathbf{w}\|^2+C\sum_i\xi_i,\qquad y_i(\mathbf{w}^T\mathbf{x}_i+b)\geq1-\xi_i,\quad\xi_i\geq0.$$

Esto equivale a una penalización cuadrática de los pesos más pérdida hinge. No vuelve curva la frontera de un SVM lineal. La penalización $L_2$ y el decaimiento de pesos coinciden en determinadas actualizaciones de SGD, pero **no son equivalentes en general para optimizadores adaptativos**; AdamW desacopla el decaimiento.



*Fig. 1.9. Minimizar la norma amplía el margen sujeto a las restricciones.*

#### 1.5.3.3. Dual y kernel trick

Para margen duro, el lagrangiano es

$$\mathcal{L}=\tfrac12\|\mathbf{w}\|^2-\sum_i\alpha_i[y_i(\mathbf{w}^T\mathbf{x}_i+b)-1].$$

La estacionariedad da $\mathbf{w}=\sum_i\alpha_i y_i\mathbf{x}_i$ y $\sum_i\alpha_i y_i=0$. El dual resulta

$$\max_{\boldsymbol{\alpha}}\sum_i\alpha_i-\frac12\sum_{i,j}\alpha_i\alpha_jy_i y_j\langle\mathbf{x}_i,\mathbf{x}_j\rangle,$$

con $\alpha_i\geq0$ y $\sum_i\alpha_i y_i=0$. En margen suave se añade $\alpha_i\leq C$.

Un kernel válido satisface $K(\mathbf{x},\mathbf{x}')=\langle\Phi(\mathbf{x}),\Phi(\mathbf{x}')\rangle$ y produce matrices de Gram semidefinidas positivas. Sustituir el producto interno permite trabajar implícitamente en otro espacio. La predicción es $\operatorname{sign}(\sum_i\alpha_i y_i K(\mathbf{x}_i,\mathbf{x})+b)$. El costo aún depende del número de muestras o vectores de soporte; el kernel no elimina la maldición de la dimensionalidad ni garantiza generalización.



*Fig. 1.10. Un kernel válido permite evitar una expansión explícita de características.*
